# HDF5 Vector/Scalar Value Checker
This notebook helps you inspect the actual values stored in your HDF5 file for columns that appear as NaN in your extracted DataFrame. Update the file path below to your target HDF5 file.

```python
import h5py
import hdf5plugin
import numpy as np
import pandas as pd
from pprint import pprint

# Path to the HDF5 file (update this as needed)
h5_path = 'test_parametric_T_seeded/ddstartup_20251113_002935_parametric_T_seeded.h5'  # <-- Change this to your actual file

# Columns to check (update as needed)
columns_to_check = [
    'I_target', 'P_DT_eq', 'P_aux', 'P_aux_DT_eq', 't_startup', 'tau_p_He3', 'unrealized_profits', 'K', 'K_el'
]

with h5py.File(h5_path, 'r') as h5:
    all_keys = list(h5.keys())
    print(f"All column names in HDF5 file ({len(all_keys)}):")
    pprint(all_keys)
    print('\n')
    print(f"Keys in HDF5 file: {all_keys}")
    # Get sol_success mask (True = 1, False = 0)
    if 'sol_success' in h5:
        sol_success = h5['sol_success'][...]
        mask = sol_success.astype(bool)
        print(f"Number of successful solutions: {np.sum(mask)} / {len(mask)}")
    else:
        print("Column 'sol_success' not found in HDF5 file.")
        mask = None
    for col in columns_to_check:
        if col in h5:
            data = h5[col][...]
            print(f"Column: {col}")
            print(f"Shape: {data.shape}")
            print(f"Type: {type(data)}")
            if mask is not None and len(data) == len(mask):
                filtered = data[mask]
                print(f"Filtered values (sol_success == True):\n{filtered}")
            elif mask is not None and data.shape[0] == len(mask):
                filtered = data[mask]
                print(f"Filtered values (sol_success == True):\n{filtered}")
            else:
                print(f"All values:\n{data}")
        else:
            print(f"Column {col} not found in HDF5 file.")
        print('-'*40)
```
